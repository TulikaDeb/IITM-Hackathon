```python
import itertools
import math

with open("scrabble list.txt", "r") as f:
    words = [line.strip().upper() for line in f if line.strip()]
```

```python
def feedback(guess, secret):
    """Compute Cats (C) and Dogs (D)"""
    D = sum(1 for i in range(4) if guess[i] == secret[i])
    C = len(set(guess) & set(secret)) - D
    return (C, D)
```

```python
# Precompute feedback for all pairs to speed up calculation
feedback_table = {}
for g in words:
    for w in words:
        feedback_table[(g, w)] = feedback(g, w)

def information_gain(guess, candidates):
    """Compute expected information gain (entropy) for a guess"""
    feedback_counts = {}
    for w in candidates:
        f = feedback_table[(guess, w)]
        feedback_counts[f] = feedback_counts.get(f, 0) + 1
    total = sum(feedback_counts.values())
    H = 0
    for count in feedback_counts.values():
        p = count / total
        H -= p * math.log2(p)
    return H
```

```python
def choose_word_max_info(candidates):
    """Pick the candidate with highest expected information gain"""
    best_word = None
    best_info = -1
    for g in candidates:
        ig = information_gain(g, candidates)
        if ig > best_info:
            best_info = ig
            best_word = g
    return best_word
```

```python
def play_game(secret, candidates):
    guesses = 0
    remaining = candidates.copy()
    while True:
        guesses += 1
        guess = choose_word_max_info(remaining)
        C, D = feedback_table[(guess, secret)]
        print(f"Guess {guesses}: {guess} -> {C}C {D}D")
        if D == 4:
            print(f"Found secret word '{secret}' in {guesses} guesses!")
            return guesses
        # Update candidate list
        remaining = [w for w in remaining if feedback_table[(guess, w)] == (C, D)]
```

```python
# Example run
secret_word = "COLD"
play_game(secret_word, words)
```

```python
import random

# Sample a subset of words, e.g., 500 words
sample_secrets = random.sample(words, 10)

total_guesses = 0
for secret in sample_secrets:
    total_guesses += play_game(secret, words)  # full candidate list

average_guesses = total_guesses / len(sample_secrets)
print("Estimated average guesses needed:", average_guesses)
```

```python
max_guesses = 0
hardest_word = ""

for secret in sample_secrets:
    guesses = play_game(secret, words)
    if guesses > max_guesses:
        max_guesses = guesses
        hardest_word = secret

print(f"Most difficult word: {hardest_word}, guesses needed: {max_guesses}")
```
